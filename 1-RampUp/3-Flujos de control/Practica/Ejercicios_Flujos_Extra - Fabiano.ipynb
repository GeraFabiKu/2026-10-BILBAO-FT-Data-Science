## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
import math
import pprint
import time

historial = []


Saludo = "Bienvenido a mi calculadora. Puede realizar operaciones de suma, resta, multiplicación, división, potencia y raíz cuadrada. También puede ver el historial de operaciones."

print(Saludo)

while True:

    seleccion_operación = input("Ingrese la operación (1=suma, 2=resta, 3=multiplicación, 4=división, 5=potencia, 6=raíz, 7=historial, 8=salir): ")

    dicc_operaciones = {
        "1": "suma",
        "2": "resta",
        "3": "multiplicación",
        "4": "división",
        "5": "potencia",
        "6": "raíz",
        "7": "historial",
        "8": "salir"
    }

    operación = dicc_operaciones.get(seleccion_operación, "operación inválida")

    match operación:
        case "suma":
            try:
                num1 = float(input("Ingrese el primer número: "))
                num2 = float(input("Ingrese el segundo número: "))
                print("Calculando...")
                time.sleep(1)
                resultado = num1 + num2
                historial.append(f"{num1} + {num2} = {resultado}")
                print(f"Resultado: {resultado}")
            except ValueError:
                print("Error: Por favor, ingrese números válidos.")
        case "resta":
            try:
                num1 = float(input("Ingrese el primer número: "))
                num2 = float(input("Ingrese el segundo número: "))
                print("Calculando...")
                time.sleep(1)
                resultado = num1 - num2
                historial.append(f"{num1} - {num2} = {resultado}")
                print(f"Resultado: {resultado}")
            except ValueError:
                print("Error: Por favor, ingrese números válidos.")
        case "multiplicación":
            try:
                num1 = float(input("Ingrese el primer número: "))
                num2 = float(input("Ingrese el segundo número: "))
                print("Calculando...")
                time.sleep(1)   
                resultado = num1 * num2
                historial.append(f"{num1} * {num2} = {resultado}")
                print(f"Resultado: {resultado}")
            except ValueError:
                print("Error: Por favor, ingrese números válidos.")
        case "división":
            try:
                num1 = float(input("Ingrese el primer número: "))
                num2 = float(input("Ingrese el segundo número: "))
                print("Calculando...")
                time.sleep(1)   
                if num2 == 0:
                    print("Error: No se puede dividir por cero.")
                else:
                    resultado = num1 / num2
                    historial.append(f"{num1} / {num2} = {resultado}")
                    print(f"Resultado: {resultado}")
            except ValueError:
                print("Error: Por favor, ingrese números válidos.")
        case "potencia":
            try:
                num1 = float(input("Ingrese la base: "))
                num2 = float(input("Ingrese el exponente: "))
                print("Calculando...")
                time.sleep(1)
                resultado = num1 ** num2
                historial.append(f"{num1} ^ {num2} = {resultado}")
                print(f"Resultado: {resultado}")
            except ValueError:
                print("Error: Por favor, ingrese números válidos.")
        case "raíz":
            try:
                num1 = float(input("Ingrese el número para calcular la raíz cuadrada: "))
                if num1 < 0:
                    print("Error: No se puede calcular la raíz cuadrada de un número negativo.")
                else:
                    print("Calculando...")
                    time.sleep(1)   
                    resultado = math.sqrt(num1)
                    historial.append(f"√{num1} = {resultado}")
                    print(f"Resultado: {resultado}")
            except ValueError:
                print("Error: Por favor, ingrese un número válido.")
        case "historial":
            pprint.pprint(historial)    
        case "salir":
            print("Saliendo del programa...")
            break

```
