```['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']```

## Ejercicio 4
Dada la siguiente lista, encuentra por el índice el valor 45 y sustitúyelo por el 0
> ```ejer_4 = [20, 47, 19, 29, 45, 67, 78, 90]```

````python
ejer_4 = [20,47,19,29,45,67,78,90]

print(ejer_4)
for i, valor in enumerate(ejer_4):
    if i == 4:
        print(f"la posición es: {i}, su valor es: {valor}")
        ejer_4[i] = 0
print(f"nueva lista actualizada: {ejer_4}")


````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3,20,3,47,19,3,29,45,67,78,90,3,3]

ejer_5.copy()

print(ejer_5.copy())

for valor in ejer_5.copy():
    if valor == 3:
        ejer_5.remove(valor) #recorremos una lista y eliminamos elementos de otra lista
print(ejer_5)
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]

diccionario = {}

for clave, valor in ejer_11:
    if clave not in diccionario:
        diccionario[clave] = []
    diccionario[clave].append(valor)

print("diccionario", diccionario)
````

## Ejercicio 15
Multiplica todos los valores del diccionario
> ```ejer_15 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}```

````python
ejer_15 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}

resultado = 1

for valor in ejer_15.values():
       resultado = resultado * valor
print(resultado)

print("=" * 48)

#Otra forma

ejer_15 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}

valores = list(ejer_15.values())

resultado = valores[0]

for valor in valores[1:]:
    print(resultado, "*", valor, "=", resultado * valor)
    resultado *= valor

print("Resultado final:", resultado)
````
