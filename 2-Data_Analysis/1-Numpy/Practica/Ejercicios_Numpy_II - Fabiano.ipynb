## Ejercicio 12
Encuentra los índices de los valores que no sean 0 del siguiente vector [1,2,0,0,4,0]. El output sería [1, 2 4]. [Encuentra el método apropiado en la documentación.](https://numpy.org/doc/) Encapsúlalo en una función, con un argumento de entrada de un vector.

````python
import numpy as np

def indices_no_cero(vector):
    return np.nonzero(vector)[0]

vector = np.array([1, 2, 0, 0, 4, 0])
print(indices_no_cero(vector))  
````

## Ejercicio 15
Crea un array cuadrado de 2D, con las dimensiones que quieras, pero que todo el borde esté compuesto por 1s, y el interior por 0s

````python
import numpy as np

array = np.zeros((5, 5), dtype=int)
array[0, :] = 1   # Borde superior
array[-1, :] = 1  # Borde inferior
array[:, 0] = 1   # Borde izquierdo
array[:, -1] = 1  # Borde derecho

print(array)

print()

array = np.ones((5, 5), dtype=int)
array[1:-1, 1:-1] = 0

print(array)
````

## Ejercicio 18
Crea una matriz aleatoria de 5x5 y normaliza sus valores. Para normalizar tienes que restar sus valores por la media(`np.mean`) y dividir el resultado por la desviación estándar(`np.std`).

````python
import numpy as np

matriz = np.random.random((5, 5))

media = np.mean(matriz)
desviacion = np.std(matriz)
matriz_normalizada = (matriz - media) / desviacion

print("Matriz inicial:\n", matriz)
print()
print("Matriz normalizada:\n", matriz_normalizada)     #normalizamos porque es una forma de estandarizar la información
````

## Ejercicio 19
1. Crea una secuencia de números que vayan del 0 al 10.
2. Crea un nuevo array, que parta del anterior, pero sustituya por el valor -1, todos los elementos cuyo valor se encuentre entre el 3 y el 8.
3. Crea un nuevo array que niegue todos los elementos entre el 3 y el 8

````python
import numpy as np

#Secuencia del 0 al 10, ambos incluidos
secuencia = np.arange(11)

#Máscara para los valores entre 3 y 8, incluidos
entre_3_y_8 = (secuencia >= 3) & (secuencia <= 8)

#Sustituir por -1
sustituidos = secuencia.copy()
sustituidos[entre_3_y_8] = -1

#Cambiar el signo de los valores entre el 3 y el 8
negados = secuencia.copy()
negados[entre_3_y_8] *= -1

print("Secuencia:", secuencia)
print()
print("Sustituidos:", sustituidos)
print()
print("Negados:", negados)


print("*"*48)
#___________________________________________________________

# Secuencia del 0 al 10, ambos incluidos
secuencia = np.arange(11)

# Sustituir por -1 los valores entre 3 y 8
sustituidos = np.where((secuencia >= 3) & (secuencia <= 8), -1, secuencia)

# Cambiar el signo de los valores entre 3 y 8
negados = np.where((secuencia >= 3) & (secuencia <= 8), -secuencia, secuencia)

print("Secuencia:", secuencia)
print()
print("Sustituidos:", sustituidos)
print()
print("Negados:", negados)


#se puede usar where
````
