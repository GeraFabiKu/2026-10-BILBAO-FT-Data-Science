# Ex1 - Filtering and Sorting Data



### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

### Step 3. Name of the max valued product

```python
chipo["item_price"] = chipo["item_price"].str.replace("$", "", regex=False).astype(float)
chipo.nlargest(1, "item_price")["item_name"]
```

### Step 4. How many products cost more than $10.00?

```python
chipo[chipo["item_price"] > 10.00][["item_name", "item_price"]]
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
precio = chipo["item_price"]

precio = pd.to_numeric(precio.str.replace("$", "", regex=False).str.strip())    #to_numeric para convertir la columna a números, similar astype

pedidos = chipo.loc[precio > 10, "order_id"].nunique()

print(pedidos)

print()

#puedo hacer tambien 
chipo.loc[precio > 10, "order_id"].nunique()
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
precio = chipo["item_price"].str.replace("$", "", regex=False).astype(float)  #astype convierte los precios de texto a números decimales

totales_por_pedido = precio.groupby(chipo["order_id"]).sum()   #uso groupby para agrupar filas x num de pedido

print((totales_por_pedido > 10).sum())


print()

#puedo hacer asi tambien porque ya tengo la columna precio
print((precio.groupby(chipo["order_id"]).sum() > 10).sum())

```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
chipo.loc[precio > 10, "order_id"].nunique()
```

### Step 5. What is the price of each item and name it unit_price. Get only item_name and unit_price

```python
chipo["unit_price"] = precio / chipo["quantity"]   #quantity para obtener el precio por unidad
chipo[["item_name", "unit_price"]].drop_duplicates()   #para quitar las filas repetidas
```

### Step 6. Sort by the name of the item

```python
chipo[["item_name", "unit_price"]].drop_duplicates().sort_values("item_name")   #para ordenar por nombre del producto
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
chipo.loc[precio.idxmax(), ["item_name", "quantity", "item_price"]]  #busca la fila del precio máximo
```

```python
chipo.assign(precio=precio).sort_values("precio", ascending=False).iloc[0][    #ordena de mayor a menor y tomar la primera fila
    ["item_name", "quantity", "item_price"]
]
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
chipo[chipo["item_name"] == "Veggie Salad Bowl"].shape[0]    #shape devuelve las dimensiones de una tabla como una pareja: (filas, columnas). El [0] selecciona el primer valor, es decir, el número de filas.

```

### Step 9. How many times did someone order more than one Canned Soda?

```python

chipo[(chipo["item_name"] == "Canned Soda") & (chipo["quantity"] > 1)].shape[0]



len(chipo.query('item_name == "Canned Soda" and quantity > 1'))  #otra forma: cuenta las filas de Canned Soda donde se pidieron más de una unidad
```
