```python
import requests
import pandas as pd

'''
requests: sirve para enviar la petición web a la API.
pandas: sirve para organizar el pronóstico en una tabla.
'''

url = "https://api.open-meteo.com/v1/forecast"

#Parametros de la consulta a la API.
params = {
    
    "latitude": 43.263,
    "longitude": -2.935,
    "current": "temperature_2m,apparent_temperature,relative_humidity_2m,wind_speed_10m",
    "daily": "weather_code,temperature_2m_max,temperature_2m_min",
    "timezone": "Europe/Madrid",
    "forecast_days": 7,
}

#Enviar la petición a la API
respuesta = requests.get(url, params=params, timeout=30)

# Comprobar si la petición tuvo éxito
if not respuesta.ok:
    print(respuesta.status_code, respuesta.text)
respuesta.raise_for_status()  # Informa de errores HTTP

#Convertir la respuesta a Python
'''La API devuelve los datos en formato JSON. 
los convierte en diccionarios y listas de Python, que se guardan en datos.'''
datos = respuesta.json()

#Mostrar las condiciones actuales
actual = datos["current"]
print(f"Bilbao: {actual['temperature_2m']} °C")
print(f"Sensación térmica: {actual['apparent_temperature']} °C")
print(f"Humedad: {actual['relative_humidity_2m']} %")
print(f"Viento: {actual['wind_speed_10m']} km/h")

#Crear y mostrar la tabla del pronóstico
pronostico = pd.DataFrame(datos["daily"])

display(pronostico)
    
```
