# Extrae los dominios de los correos con una expresión regular

```python
import re

emails = '''
miguellc@thebridgeschool.tech
miguel.cruz@uc3m.es
miguel-12-2000@mi-correo.net
miguellc@s@thebridgeschool.es
'''
```

Ejemplo:
* @thebridgeschool.tech
* @uc3m.es
* @mi-correo.net
* @thebridgeschool.es

```python
dominio = re.findall(r'@[\w.-]+\.[A-Za-z]{2,}', emails)
dominio

'''@ — busca el carácter @.
[\w.-]+ — acepta uno o más caracteres de palabra (\w: letras, dígitos y _), puntos o guiones.
\. — busca un punto literal. La barra invertida evita que . signifique «cualquier carácter».
[A-Za-z]{2,} — acepta dos o más letras inglesas, mayúsculas o minúsculas.'''   
```

```python
pattern = re.compile(r'@([a-zA-Z0-9.-]+([^@]\b))')
pattern.findall(emails)

```
